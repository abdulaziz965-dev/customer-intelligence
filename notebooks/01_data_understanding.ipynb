## Data Cleaning

### Converting TotalCharges to Numeric

The TotalCharges column contains numerical values
stored as strings. We will convert it into a numeric
column and identify invalid or missing entries.

```python
df_clean=df.copy()

df_clean["TotalCharges"]=df_clean["TotalCharges"].str.strip()

df_clean["TotalCharges"]=pd.to_numeric(df_clean["TotalCharges"], errors="coerce")

print(df_clean["TotalCharges"].dtypes)

print("Missing Values:", df_clean["TotalCharges"].isnull().sum())
```

```python
missing_rows=df_clean[df_clean["TotalCharges"].isnull()]

display(missing_rows[["customerID", "TotalCharges", "MonthlyCharges", "Churn"]])
```

```python
print("Missing records:", len(missing_rows))

print("\nTenure distribution:")
print(missing_rows["tenure"].value_counts())

print("\nAverage tenure of all customers:",
      df_clean["tenure"].mean())
```

```python
# Replace missing TotalCharges with 0
df_clean["TotalCharges"] = (
    df_clean["TotalCharges"].fillna(0)
)

# Verify
print("Missing TotalCharges:",
      df_clean["TotalCharges"].isnull().sum())

print("Data type:",
      df_clean["TotalCharges"].dtype)
```

```python
# Check all missing values
print("Missing values per column:")
print(df_clean.isnull().sum())

# Check duplicates
print("\nDuplicate records:",
      df_clean.duplicated().sum())

# Check dataset dimensions
print("\nDataset shape:", df_clean.shape)

# Check TotalCharges statistics
print("\nTotalCharges statistics:")
print(df_clean["TotalCharges"].describe())
```

```python
df_clean.to_csv(
    "../data/processed/cleaned_customer_data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
